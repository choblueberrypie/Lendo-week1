# Week 1 — Environment Hello World

This notebook verifies that the reproducible Python environment can import every library required for the Week 1 market-data workflow.

```python
from __future__ import annotations

import sys

import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import pyarrow
import scipy
import seaborn as sns
import sklearn
import statsmodels
import statsmodels.api as sm

versions = {
    "python": sys.version.split()[0],
    "numpy": np.__version__,
    "pandas": pd.__version__,
    "scipy": scipy.__version__,
    "pyarrow": pyarrow.__version__,
    "statsmodels": statsmodels.__version__,
    "scikit-learn": sklearn.__version__,
    "matplotlib": matplotlib.__version__,
    "seaborn": sns.__version__,
}

assert sm.__name__ == "statsmodels.api"

for package, version in versions.items():
    print(f"{package}: {version}")

print("All required libraries imported successfully.")
```

```python
sample = pd.DataFrame(
    {
        "date": pd.date_range("2026-01-01", periods=3, freq="D"),
        "ticker": ["SPY", "SPY", "SPY"],
        "adj_close": [100.0, 101.0, 102.01],
        "volume": [1_000_000, 1_100_000, 1_050_000],
    }
)

sample["simple_return"] = sample["adj_close"].pct_change()

expected_second_return = 0.01
actual_second_return = sample.loc[1, "simple_return"]
assert np.isclose(actual_second_return, expected_second_return)

fig, ax = plt.subplots(figsize=(5, 2.5))
sns.lineplot(data=sample, x="date", y="adj_close", marker="o", ax=ax)
ax.set_title("Environment smoke test")
plt.tight_layout()
plt.close(fig)

sample
```
